```python
import math

import math
import matplotlib.pyplot as plt

legs = [(100, 0), (100, 90), (100, 180), (100, 270)]

e = 0
n = 0
perimeter = 0
points = [(0, 0)]

for dist, bearing in legs:
  b = math.radians(bearing)
  e = e + dist * math.sin(b)
  n = n + dist * math.cos(b)
  perimeter = perimeter + dist
  points.append((e, n))

error = math.hypot(e, n)
print("Closing error (m):", round(error, 3))
print("Perimeter (m):", perimeter)
print("Precision: 1 in", round(perimeter / error))

xs = [p[0] for p in points]
ys = [p[1] for p in points]
plt.plot(xs, ys, marker="o")
plt.axis("equal")
plt.title("Survey traverse")
plt.show()
```
